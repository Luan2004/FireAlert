```python
import yaml

data_yaml = "/kaggle/working/FireSmokeDataset-1/data.yaml"

data = {
    "path": "/kaggle/working/FireSmokeDataset-1",   # thư mục gốc dataset
    "train": "/kaggle/working/FireSmokeDataset-1/train/images",
    "val":   "/kaggle/working/FireSmokeDataset-1/valid/images",
    "test":  "/kaggle/working/FireSmokeDataset-1/test/images",       # bỏ nếu không có
    "nc": 3,                       # số lượng class
    "names": ["fire","other", "smoke" ]     # tên các class
}

with open(data_yaml, "w") as f:
    yaml.dump(data, f, sort_keys=False)

print(open(data_yaml).read())
```

# Train

```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")
results = model.train(
    data=data_yaml,
    epochs=400,
    imgsz=640,
    batch=64,
    device="0,1",
    workers=8,
    project="/kaggle/working/fire-8",
    name="yolov8n",
    pretrained=True,
    patience=30,
    cache=False,
    optimizer="AdamW",
    lr0=0.00005,
    lrf=0.01
)
```
